Temos um problema de recomendação com n usuários a \in {1, ... ,n} e m itens i € {1, ... ,m}. Usaremos os rótulos {-1,1} para representar a avaliação alvo (desgosta,gosta). Cada usuário provavelmente fornecerá feedback para apenas um pequeno subconjunto dos itens possíveis, e por isso devemos restringir os modelos para evitar overfitting. Nosso objetivo neste problema é entender como um modelo simples de rede neural se aplica a esse problema e quais são as restrições do modelo.

Unidades de Entrada
Considere a rede neural simples com 1 camada oculta representada na figura acima. Usamos uma unidade de entrada para cada usuário (os nós na coluna à esquerda) e para cada item (os nós na linha superior); portanto, no total, há n + m unidades de entrada.

Ao fazer a previsão para uma entrada selecionada (a, i), apenas a a^a unidade de entrada do usuário e i^a unidade de entrada do item estao ativas (ou seja, definidas com o valor 1); todas as outras entradas são definidas para 0 e não afetarão as previsões. Em outras palavras, apenas os pesos de saida dessas duas unidades são relevantes para prever o rótulo (-1 ou 1) para a entrada (a, i).

Unidades Ocultas
Usuario a tem dois pesos de saída, Ua1 e Ua2 , e o item tem dois pesos de saida, Vi1 e Vi2 . Esses pesos são alimentados como entradas para as duas unidades ocultas no modelo. As unidades ocultas avaliam

z1 = Ua1 +Vi1, f(z1)= max{0,z1}
z2 = Ua2+Vi2, f(z2) = max{0,z2}.

Saida
Assim, para a entrada (a, i), nossa rede gera
F(a,i;𝜃) = W1f(z1)+W2f(z2)+W0

onde 𝜃 denota todos os pesos U, V, e W. Finalmente, uma função de sinal é aplicada a F (a, i; 𝜃) para a classificação.

Na notação vetorial, cada usuário a tem um vetor bidimensional de pesos de saída ua = [Ua1, Ua2]^T, e, de forma similar, cada item i tem um vetor bidimensional de pesos de saida vi = [Vi1, Vi2]^T . A entrada recebida pelas unidades ocultas e representada como o vetor

Z=[z1,\2]^T=ua+vi

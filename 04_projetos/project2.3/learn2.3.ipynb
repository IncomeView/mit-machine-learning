### ⭐ Variáveis fundamentais da rede neural

#### 1. **V** — pesos da camada oculta → saída  
$
u = Vh
$

**O que é:**  
- matriz (ou vetor) de pesos que conecta **cada neurônio da camada oculta** ao **neurônio de saída**.

**Dimensão:**  
- se a camada oculta tem 3 neurônios, então  
  $
  V = [v_1\ v_2\ v_3]
  $

**Interpretação geométrica:**  
- V define **como cada feature aprendida pela camada oculta contribui para a saída**.  
- É literalmente a inclinação da combinação linear final.

**No backprop:**  
$
\frac{\partial C}{\partial V} = \delta_{\text{out}} \cdot h^T
$

---

#### 2. **u** — pré‑ativação da camada de saída  
$
u = Vh
$

**O que é:**  
- o valor **antes da função de ativação** da saída.  
- como a ativação é identidade,  
  $
  \hat{y} = u
  $

**Interpretação geométrica:**  
- é o ponto em que a rede ainda é **linear**.  
- é a soma ponderada das ativações da camada oculta.

**No backprop:**  
$
\delta_{\text{out}} = \frac{\partial C}{\partial u} = \hat{y} - y
$


**Deriivada de $u=Vh$**

$$
u = Vh \begin{bmatrix} v_1 & v_2 & v_3 \end{bmatrix} 
		\begin{bmatrix} h_1 \\ h_2 \\ h_3 \end{bmatrix}
			 = v_1h_1 + v_2h_2 + v_3h_3 
			 \quad \Rightarrow \quad  \frac{\partial_u}{\partial_h} =  ?
$$

$$ \frac{\partial_u}{\partial h_{1}} = v_1, \quad \frac{\partial_u}{\partial h_{2}} = v_2, \quad \frac{\partial_u}{\partial h_{3}} = v_3
\qquad \Rightarrow \qquad \frac{\partial_u}{\partial_h} = \begin{bmatrix} v_1 \\ v_2 \\ v_3 \end{bmatrix} = V^T $$



---

#### 3. **C** — função de custo (erro quadrático)  
$
C = \frac{1}{2}(y - \hat{y})^2
$

**O que é:**  
- mede o erro entre o valor previsto e o valor real.  
- quanto maior o erro, maior o custo.

**Interpretação geométrica:**  
- é a distância quadrática entre o ponto previsto e o ponto real.  
- cria uma superfície convexa que o gradiente descendente pode minimizar.

**No backprop:**  
$
\frac{\partial C}{\partial \hat{y}} = \hat{y} - y
$

Esse termo aparece em toda rede neural com MSE.

---

### ⭐ Outras variáveis essenciais (para completar o quadro)

#### 4. **W** — pesos da entrada → camada oculta  
$
z = Wx + b
$

**Interpretação:**  
- define como cada pixel (ou feature) influencia cada neurônio oculto.

---

#### 5. **b** — vieses da camada oculta  
$
z = Wx + b
$

**Interpretação:**  
- desloca o hiperplano da ReLU.  
- controla onde a função muda de 0 para linear.

---

#### 6. **z** — pré‑ativação da camada oculta  
$
z = Wx + b
$

**Interpretação:**  
- é o ponto onde a ReLU decide:  
  - se o neurônio **ativa** (z > 0)  
  - ou **apaga** (z ≤ 0)

---

#### 7. **h** — ativação da camada oculta  
$
h = \text{ReLU}(z)
$

**Interpretação:**  
- são as features aprendidas pela rede.  
- cada neurônio representa uma região linear diferente.

---

#### 8. **δ_out** — erro da camada de saída  
$
\delta_{\text{out}} = \hat{y} - y
$

**Interpretação:**  
- é o quanto a saída está errada.  
- é o ponto de partida do backprop.

---

#### 9. **δ_hidden** — erro da camada oculta  
$
\delta_{\text{hidden}} = (V^T \delta_{\text{out}}) \odot \text{ReLU}'(z)
$

**Interpretação:**  
- diz quanto cada neurônio oculto contribuiu para o erro final.  
- controla o aprendizado da primeira camada.

| | | $ \qquad 11.3\to\quad V \leftarrow V - \eta \cdot \frac{\partial C}{\partial V}  $ | | |

**Uma batch é um mini conjunto de treinamento, processado ao mesmo tempo**
- Em vez de treinar com **um exemplo por vez**, você treina com **32 exemplos ao mesmo tempo**.
- Isso é feito para
	- acelera o treinamento
	- estabiliza o gradiente
	- melhora a convergência
	- permite paralelização (GPU)
	- reduz ruído estatístico

**Matemática**
- Antes: $ z = Wx + b $
- Agora: $ Z = XW^T + b $
- **Onde:**
	- $X$ é ima matriz $ 32 x 784 $
	- Cada linha é uma imagem
	- o forward é feito para** todos os 32 exemplos ao mesmo tempo**
- **O gradiente também é calculado**
	- $ \frac{\partial C}{\partial W} = \delta \cdot x^T $
	- $ \frac{\partial C}{\partial W} = \sum_{i=1}^{32} \delta_i \cdot x_i^T : \quad$ o que o **Pytorch** faz
	- **Ou seja**
		- o gradiente é a soma dos gradientes de cada exemplo
		- isso deixa o aprendizado mais estável

O treinamento com 32 exemplos **utiliza a mesma matemática** do treinamento de 1 por vez, só que:
- com matrizes
- com soma de gradientes
- com mais estabilidade

````python
train_batches = batchify_data(X_train, y_train, batch_size)
dev_batches = batchify_data(X_dev, y_dev, batch_size)
test_batches = batchify_data(X_test, y_test, batch_size)
````

**Relação Momentum e lr**
$$ v_t = \mu v_{t-1} + \eta \frac{\partial C}{\partial W} $$

**Intuição simples**
- **Momentum: funciona como a memoria do gradiente passado**
	- acelerar o aprendizado
	- suavizar o gradiente
	- evitar ficar preso em mínimos locais
	- evitar oscilações
	- **Imagine que o gradiente aponta para baixo da montanha, mas o terreno é cheio de pequenas ondulações.**
		- **Sem momentum:**
			- o gradiente muda muito
			- o passo fica “tremendo”
			- o aprendizado fica lento
		- **Com momentum:**
			- você acumula velocidade
			- o passo fica mais suave
			- você desce a montanha mais rápido


- **lr (learning rate | $\eta$): é o número que controla o tamanho do passo que o otimizador dá na direção do gradiente**
	-	Imagine que você está descendo uma montanha (minimizando a perda).
		-	✔ Se a taxa de aprendizado é grande demais
			- Você dá passos enormes:
			- pula o vale
			- oscila
			- explode o gradiente
			- a perda não converge
		- ✔ Se a taxa de aprendizado é pequena demais
			- Você dá passos minúsculos:
			- demora muito para descer
			- pode ficar preso em mínimos locais
			- o treinamento fica lento
			- a acurácia fica baixa
		- ✔ Se a taxa de aprendizado é boa
			- Você desce:
			- rápido
			- suave
			- sem oscilações
			- com boa convergência

### 9. Redes Neurais Convolucionais (CNN) | PyTorck

As CNNs são uma **evolução das redes totalmente conectadas**.
- Enquanto o MLP recebe a imagem achatada como um vetor de 784 números, 
- a CNN preserva a estrutura espacial da imagem: $z \in R^{1x28x28} $

Ou seja:
- 1 canal (imagem em tns de cinza)
- 28 x 28 pixels
- sem achatamento incial

> Isso permite que a rede aprenda padrões locais, como bordas, curvas e partes de dígitos — algo impossível para o MLP.


````python
nn.Conv2d(1, 32, (3, 3))
````

| **Camada Convolucional**| $$ Z_{i,j}^{(k)} = \sum_{u=1}^3 \sum_{v=1}^3 W_{u,v}^{(k)} \cdot x_{i+u, j+v} + b{(k)} $$ |
|-|-|

Onde:
- k é o índice do filtro (32 filtros)
- cada filtro aprende um padrão diferente (bordas horizontais, verticais, curvas etc.)
- o resultado é um mapa de características de tamanho $26\text{x}26$
	- Estamos considerando $ \text{padding = 0} $ padrão do PyTorch,
	- e sem padding, o kernel não consegue "encaixar" nas bordas

Para uma entrada de tamanho $ 𝐻 \text{×}𝑊$, kernel $𝐾$, padding $𝑃$, stride $𝑆$: 
- $\Rightarrow \quad \text{output size} = \frac{H - K + 2P}{S}+1 $
- Neste caso:
	- $H=28$
	- $K=3$
	- $P=0$
	- $S=1$
- $\Rightarrow \quad \text{output} = \frac{28 - 3 + 0}{1}+1 = 26 \Rightarrow \quad 26\text{x}26 $

Intuição:  
- Cada filtro “varre” a imagem e responde fortemente quando detecta o padrão que aprendeu.

````python
nn.ReLU()
````

**Não-linearidade ReLU**
- Mantém apenas ativações positivas, introduzindo não-linearidade: $h = max(0, z)$
- $\text{ReLU}'(x) = \begin{cases} x, & x > 0 \\ 0, & x \le 0 \end{cases} $

````python
nn.MaxPool2d((2, 2))
````

**Max Pooling**

- Reduz a dimensção espacial pela metade: $ 26 \text{ x } 26 \to 13 \text{ x } 13 $
	- kernel_size = 2
	- stride = 2, $\quad\to$ motivo pela diminuição espacial no deslocamento 
- Matematicamente: $ h{i,j} = \max\{{z_{2i,2j}, z_{2i + 1, 2j}, z_{2i,2j+1}, z_{2i +1,2j +1} }\} $

Intuição:  
> Mantém apenas a informação mais importante de cada região.

**Flatten() - achatar o tensor**

- Ele pega um tensor 3D: $ \text{shape} = (C, H, W)$
- e transforma em um vetor 1D: $ \text{shape} = (C \cdot H \cdot W)$
- Transformando:  $64 \text{ x } 5 \text{ x }  5 \to 1600 $ 

Matemática
- Se o tensor é: $X \in R^{C\text{x}H\text{x}W}$
- O Flatten produz: $X \in R^{CHW}$
- Com a regra: $x_{(c \cdot HW) + (h \cdot W) + w} = X_{c, h, w}$


> Esse vetor alimenta as camadas totalmente conectadas.

Porque:
- Convoluções trabalham com tensores 3D (canal × altura × largura)
- Fully Connected (Linear) trabalha com vetores 1D
- A camada Linear espera algo assim: $ \to\quad x = \begin{bmatrix} x_1 \\ x_2 \\ \vdots \\ x_{1600} \end{bmatrix} $

Intuição visual
- Imagine que você tem 64 “imagens” de 5×5 (os feature maps).
- O Flatten pega tudo isso e coloca em uma única linha:

```codigo
[mapa1_pixel1, mapa1_pixel2, ..., mapa64_pixel25]
```

- É literalmente “achatar” o tensor.
- Esse vetor contém todas as ativações dos filtros convolucionais.
- Cada número representa:
	- uma borda detectada
	- uma curva
	- uma textura
	- uma parte do dígito
	- uma combinação de padrões

````python
nn.Linear(1600, 128)
````

**A ponte entre CNN e MLP**

- A camada Linear faz: $ℎ = 𝑊𝑥 + 𝑏$

Onde:
- $ x \in R^{1600} $
- $ W \in R^{128\text{x}1600} $
- $ b \in R^{128} $
- $ h \in R^{128} $


✔ Intuição
- Essa camada aprende como combinar os padrões detectados pela CNN.
- Exemplo:
	- filtro 1 detecta borda esquerda
	- filtro 2 detecta curva superior
	- filtro 3 detecta parte inferior do dígito

A Linear aprende que:
> “Se esses três padrões aparecem juntos, isso parece um 5.”

Ela é a primeira camada “sem visão espacial”, que só combina informações.

````python
nn.Dropout(0.5)
````

**Evita overfitting e força generalização**

- O dropout implementa: $h_i = \begin{cases} 0, & \text{com probabilidade } 0.5 \\ h_i, & \text{caso contrário} \end{cases} $

Ou seja:
- metade dos neurônios é desligada aleatoriamente
- a rede não pode depender de um único caminho
- ela aprende representações mais robustas
- reduz overfitting
- melhora generalização

✔ Intuição
> Dropout é como “treinar várias redes diferentes ao mesmo tempo”.

Cada forward passa por uma rede ligeiramente diferente.

````python
nn.Linear(128, 10)
````

**A camada de decisão final**

- Essa camada produz os logits: $u = Vh + c$


Onde:
- $ x \in R^{128} $
- $ W \in R^{10\text{x}128} $
- $ b \in R^{10} $
- $ h \in R^{10} $

Assim:
- Cada componente de $𝑢$ representa a “força” da rede para cada classe: $ u_0, u_1, ..., u_9$  
- O PyTorch aplica automaticamente: $\text{CrossEntropyLoss} = \text{LogSoftmax} + \text{NLLLoss}$  
- Então você não precisa colocar softmax manualmente.

✔ Intuição
Essa camada responde:

> “Com base em tudo o que aprendi, qual dígito isso parece ser?”

````python
nn.Linear(1600, 128)
nn.Dropout(0.5)
nn.Linear(128, 10)
````

**Camada Linear + Dropout**

- $u = Vh + c$

Onde:
- $h \in R^{128}$
- $u \in R^{10}$ (logits)
- O dropout implementa: $h_i = \begin{cases} 0, & \text{com probabilidade } 0.5 \\ h_i, & \text{caso contrário} \end{cases} $

Intuição:  
> Evita overfitting e força a rede a aprender representações mais robustas.

**Treinamento**

O treinamento da CNN segue exatamente o mesmo fluxo do MLP:
- forward
- cálculo da loss (CrossEntropy)
- backpropagation
- atualização via SGD + momentum

A diferença é que agora o gradiente flui por:
- filtros convolucionais
- mapas de características
- pooling
- camadas fully connected

O PyTorch calcula tudo automaticamente.

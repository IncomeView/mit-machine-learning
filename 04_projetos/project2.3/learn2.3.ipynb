### ⭐ Variáveis fundamentais da rede neural

#### 1. **V** — pesos da camada oculta → saída  
$
u = Vh
$

**O que é:**  
- matriz (ou vetor) de pesos que conecta **cada neurônio da camada oculta** ao **neurônio de saída**.

**Dimensão:**  
- se a camada oculta tem 3 neurônios, então  
  $
  V = [v_1\ v_2\ v_3]
  $

**Interpretação geométrica:**  
- V define **como cada feature aprendida pela camada oculta contribui para a saída**.  
- É literalmente a inclinação da combinação linear final.

**No backprop:**  
$
\frac{\partial C}{\partial V} = \delta_{\text{out}} \cdot h^T
$

---

#### 2. **u** — pré‑ativação da camada de saída  
$
u = Vh
$

**O que é:**  
- o valor **antes da função de ativação** da saída.  
- como a ativação é identidade,  
  $
  \hat{y} = u
  $

**Interpretação geométrica:**  
- é o ponto em que a rede ainda é **linear**.  
- é a soma ponderada das ativações da camada oculta.

**No backprop:**  
$
\delta_{\text{out}} = \frac{\partial C}{\partial u} = \hat{y} - y
$


**Deriivada de $u=Vh$**

$$
u = Vh \begin{bmatrix} v_1 & v_2 & v_3 \end{bmatrix} 
		\begin{bmatrix} h_1 \\ h_2 \\ h_3 \end{bmatrix}
			 = v_1h_1 + v_2h_2 + v_3h_3 
			 \quad \Rightarrow \quad  \frac{\partial_u}{\partial_h} =  ?
$$

$$ \frac{\partial_u}{\partial h_{1}} = v_1, \quad \frac{\partial_u}{\partial h_{2}} = v_2, \quad \frac{\partial_u}{\partial h_{3}} = v_3
\qquad \Rightarrow \qquad \frac{\partial_u}{\partial_h} = \begin{bmatrix} v_1 \\ v_2 \\ v_3 \end{bmatrix} = V^T $$



---

#### 3. **C** — função de custo (erro quadrático)  
$
C = \frac{1}{2}(y - \hat{y})^2
$

**O que é:**  
- mede o erro entre o valor previsto e o valor real.  
- quanto maior o erro, maior o custo.

**Interpretação geométrica:**  
- é a distância quadrática entre o ponto previsto e o ponto real.  
- cria uma superfície convexa que o gradiente descendente pode minimizar.

**No backprop:**  
$
\frac{\partial C}{\partial \hat{y}} = \hat{y} - y
$

Esse termo aparece em toda rede neural com MSE.

---

### ⭐ Outras variáveis essenciais (para completar o quadro)

#### 4. **W** — pesos da entrada → camada oculta  
$
z = Wx + b
$

**Interpretação:**  
- define como cada pixel (ou feature) influencia cada neurônio oculto.

---

#### 5. **b** — vieses da camada oculta  
$
z = Wx + b
$

**Interpretação:**  
- desloca o hiperplano da ReLU.  
- controla onde a função muda de 0 para linear.

---

#### 6. **z** — pré‑ativação da camada oculta  
$
z = Wx + b
$

**Interpretação:**  
- é o ponto onde a ReLU decide:  
  - se o neurônio **ativa** (z > 0)  
  - ou **apaga** (z ≤ 0)

---

#### 7. **h** — ativação da camada oculta  
$
h = \text{ReLU}(z)
$

**Interpretação:**  
- são as features aprendidas pela rede.  
- cada neurônio representa uma região linear diferente.

---

#### 8. **δ_out** — erro da camada de saída  
$
\delta_{\text{out}} = \hat{y} - y
$

**Interpretação:**  
- é o quanto a saída está errada.  
- é o ponto de partida do backprop.

---

#### 9. **δ_hidden** — erro da camada oculta  
$
\delta_{\text{hidden}} = (V^T \delta_{\text{out}}) \odot \text{ReLU}'(z)
$

**Interpretação:**  
- diz quanto cada neurônio oculto contribuiu para o erro final.  
- controla o aprendizado da primeira camada.

## codigo

````python
def train(self, x1, x2, y):

    ### Forward propagation ###
    input_values = np.matrix([[x1], [x2]])  # 2x1

    # Hidden layer
    hidden_layer_weighted_input = (
        self.input_to_hidden_weights * input_values + self.biases
    )  # 3x1
    # aplica ReLU elemento a elento na matriz
    relu = np.vectorize(rectified_linear_unit)
    # é o h | h = ReLU(z) = max(0, z)
    hidden_layer_activation = relu(hidden_layer_weighted_input)  # 3x1

    # Output layer
    # output é u | u = V * h
    output = self.hidden_to_output_weights * hidden_layer_activation  # 1x1
    # É a ativação da camada de saída, que neste caso é a função identidade
    activated_output = output_layer_activation(output)  # identity

    ### Backpropagation ###

    # Output error
    output_layer_error = activated_output - y  # 1x1

    # Hidden layer error
    relu_deriv = np.vectorize(rectified_linear_unit_derivative)
    hidden_layer_error = np.multiply(
        np.transpose(self.hidden_to_output_weights) * output_layer_error,
        relu_deriv(hidden_layer_weighted_input),
    )  # 3x1

    # Gradients
    bias_gradients = hidden_layer_error  # 3x1
    hidden_to_output_weight_gradients = output_layer_error * np.transpose(
        hidden_layer_activation
    )  # 1x3
    input_to_hidden_weight_gradients = hidden_layer_error * np.transpose(
        input_values
    )  # 3x2

    ### Gradient descent update ###
    self.biases -= self.learning_rate * bias_gradients
    self.input_to_hidden_weights -= (
        self.learning_rate * input_to_hidden_weight_gradients
    )
    self.hidden_to_output_weights -= (
        self.learning_rate * hidden_to_output_weight_gradients
    )
````
